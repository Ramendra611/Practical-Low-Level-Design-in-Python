## Builder Pattern
    build the object in steps, with a clean API, and only produce the final object once all configuration is done.

```python
class Pizza:
    def __init__(self):
        self.size = None
        self.crust = None
        self.cheese = None
        self.toppings = []
        self.extra_cheese = False

    def __repr__(self):
        return (f"Pizza(size={self.size}, crust={self.crust}, "
                f"cheese={self.cheese}, toppings={self.toppings}, "
                f"extra_cheese={self.extra_cheese})")


class PizzaBuilder:
    def __init__(self):
        self.pizza = Pizza()

    def set_size(self, size):
        self.pizza.size = size
        # Returning `self` is what makes the fluent interface work: each setter
        # hands back the SAME builder, so the next method can be called directly
        # on the result — `.set_size(...).set_crust(...)`. Without this return,
        # each call would yield None and chaining would break on the second dot.
        return self

    def set_crust(self, crust):
        self.pizza.crust = crust
        return self

    def set_cheese(self, cheese):
        self.pizza.cheese = cheese
        return self

    def add_topping(self, topping):
        self.pizza.toppings.append(topping)
        return self

    def with_extra_cheese(self):
        self.pizza.extra_cheese = True
        return self

    def build(self) -> Pizza:
        return self.pizza


# Usage — fluent, readable
pizza = (PizzaBuilder()
         .set_size("large")
         .set_crust("thin")
         .set_cheese("mozzarella")
         .add_topping("olives")
         .add_topping("mushrooms")
         .with_extra_cheese()
         .build())

print(pizza)
```

#### Simple Factory

```python
from abc import ABC, abstractmethod

class Notification(ABC):
    @abstractmethod
    def send(self, msg):
        pass

class Email(Notification):
    def send(self, msg):
        print("Sending email. ")

class Sms(Notification):
    def send(self, msg):
        print("Sending Sms. ")

class Whatsapp(Notification):
    def send(self, msg):
        print("Sending whatsapp message. ")

class Push(Notification):
    def send(self, msg):
        print("sending push notification")

class NotificationProvider: # factory

    @staticmethod
    def provide_notification(notification_type): # violating the OCP
        if notification_type == "email":
            return  Email()
        
        if notification_type == "sms":
            return Sms()
        
        if notification_type == "whatsapp":
            return Whatsapp()

        if notification_type == "push":
            return Push()
            
        else:
            raise Exception("invalid type!")


msg = "hello user"

# create the object and then send the message
notification = NotificationProvider().provide_notification("push")
notification.send(msg)
```

 - The factory centralizes creation logic. SO object creation is de-coupled from the business logic

```python
class Notification(ABC):
    @abstractmethod
    def send(self, msg):
        pass

class Email(Notification):
    def send(self, msg):
        print("Sending email. ")

class Sms(Notification):
    def send(self, msg):
        print("Sending Sms. ")

class Whatsapp(Notification):
    def send(self, msg):
        print("Sending whatsapp message. ")

class Push(Notification):
    def send(self, msg):
        print("sending push notification")


class NotificationService(ABC):
    
    @abstractmethod
    def provide_notification() -> Notification:
        '''
        this will be implemented by the subclass (concrete class )
        '''
        pass

    def notify(self, msg):
        ## create the notification object and then send the message using the object
        notification = self.provide_notification()
        notification.send(msg)

class EmailService(NotificationService):
    def provide_notification(self) -> Notification: 
        return Email() # return email object

class SmsService(NotificationService):
    def provide_notification(self) -> Notification: 
        return Sms()

class WhatsappService(NotificationService):
    def provide_notification(self) -> Notification: 
        return Whatsapp()

class PushNotificationService(NotificationService):
    def provide_notification(self) -> Notification: 
        return Push()
    

msg = "hello user"
service = PushNotificationService()
service.notify(msg)
```

#### Abstract Factory
    - Factory of factory

```python
## we are trying to create some GUI ( Button, Window, Checkbox ) ---> we want to create for different OS ( Ios , Windows, Mac, Linux )
```

```python
class Button(ABC):
    @abstractmethod
    def render(self):
        pass

class Checkbox(ABC):
    @abstractmethod
    def render(self):
        pass

class MacButton(Button):
    def render(self):
        print("I am a button in MacOS")

class WindowsButton(Button):
    def render(self):
        print("I am a button in windowsOS")


class MacCheckbox(Button):
    def render(self):
        print("I am a checkbox in MacOS")

class WindowsCheckbox(Button):
    def render(self):
        print("I am a checkbox in windowsOS")


class UIFactory(ABC):

    def create_button(self) -> Button:
        pass

    def create_checkbox(self) -> Checkbox:
        pass


class MacUIFactory(UIFactory):
    def create_button(self):
        print("Creating mac button")
        return MacButton()

    def create_checkbox(self):
        return MacCheckbox()


class WindowsUIFactory(UIFactory):
    def create_button(self):
        print("Creating windows button")
        return WindowsButton()

    def create_checkbox(self):
        return WindowsCheckbox()


## task is to build a UI ( select which UI factory(windows/mac) --> goto the UI factory and call the button and checkbox method )
def build_ui(factory:  UIFactory ):
    button = factory.create_button()
    checkbox = factory.create_checkbox()

    ## render the ui elements
    button.render() 
    checkbox.render()



# build_ui(MacUIFactory())
build_ui(WindowsUIFactory())    
```

    - Extending the idea of the factory method
    - It acts like a factory of factory
    - Difference between the Factory method and abstract factory:
            - factory method is creating one object but abstract factory is creating a family of objects
            - abstract factory method takes another factory as a parameter. We are swapping entire factory. 
            - factory method is related to "what object is needed?" and abstract factory is related to "what family of object is needed?"


## Prototype Pattern
    - creational pattern that comes into picture when the object creation process is expensive and the object is created by
        cloning an existing object. 
    - we also use a prototype to create an object

    - how copy works in python --> shallow copy and deep copy?

```python
import copy

class Document:
    def __init__(self, title, body, tags=None, metadata=None):
        self.title = title
        self.body = body
        self.tags = tags or []
        self.metadata = metadata or {}

    def clone(self):
        return copy.deepcopy(self)

    def shallow_clone(self):
        return copy.copy(self)

    def __repr__(self):
        return f"Document({self.title!r}, tags={self.tags}, metadata={self.metadata})"


prototype = Document("Blog Template", "Introduction...",
                     tags=["draft"], metadata={"author": "Amit"})

# Deep clone — independent
deep = prototype.clone()
deep.tags.append("published")
deep.metadata["reviewed"] = True
print("Original after deep clone mutation:", prototype)
# Original is unchanged: tags=['draft'], metadata={'author': 'Amit'}
print("Deep clone:", deep)

# Shallow clone — dangerous!
shallow = prototype.shallow_clone()
shallow.tags.append("shared!")
print("Original after shallow clone mutation:", prototype)
# Original.tags is now ['draft', 'shared!'] — mutation leaked
```
